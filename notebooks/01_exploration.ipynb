M2 — Exploration des données

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

BASE_DIR = Path.cwd().parent
DATA_DIR = BASE_DIR / "data" / "input"
customers = pd.read_csv(DATA_DIR / "customers_data - customers_data.csv")
products = pd.read_csv(DATA_DIR / "products_data - products_data.csv")
sales = pd.read_csv(DATA_DIR / "sales_data - sales_data.csv")
marketing = pd.read_csv(DATA_DIR / "marketing_data - marketing_data.csv")

print("Customers :", customers.shape)
print("Products  :", products.shape)
print("Sales     :", sales.shape)
print("Marketing :", marketing.shape)

print("Customers:")
print(customers.head())
print("Products:")
print(products.head())
print("Sales:")
print(sales.head())
print("Marketing:")
print(marketing.head())

print("Customers Info:")
customers.info()
print("Products Info:")
products.info()
print("Sales Info:")
sales.info()
print("Marketing Info:")
marketing.info()
```

```python
#Nettoyage de données
print("Customers:")
print(customers.isnull().sum())

print("\nProducts:")
print(products.isnull().sum())

print("\nSales:")
print(sales.isnull().sum())

print("\nMarketing:")
print(marketing.isnull().sum())
```

```python
#V"rifications des doublons
customers.duplicated().sum()
products.duplicated().sum()
sales.duplicated().sum()
marketing.duplicated().sum()


#Valeurs uniques des variables catégorielles
print("Unique values in Customers:")
print(customers.nunique())
print("Unique values in Products:")
print(products.nunique())
print("Unique values in Sales:")
print(sales.nunique())
print("Unique values in Marketing:")
print(marketing.nunique())

print("Customers describe:")
print(customers.describe())
print("Products describe:")
print(products.describe())
print("Sales describe:")
print(sales.describe())
print("Marketing describe:")
print(marketing.describe())
```
